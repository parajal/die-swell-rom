```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent
sys.path.insert(0, str(ROOT))
from romlab import FOM, POD, OfflinePhase, OnlinePhase, Plot

files_train = ["snapshots.txt", "c-trace.txt",  "parameters.txt"]
files_test  = ['snapshots_test.txt', 'c-trace_test.txt', 'parameters_test.txt']

# Full Order Model

fom = FOM(
    filenames_train=files_train,
    filenames_test=files_test,
    data_folder=ROOT / "data/2d/lambda-beta-time",
    mesh_file="mesh.vtk", FIELDS = ("velocity", "c-trace"),
    param_cols=[0, 1, 3], seed = 42)

fom.info()
```

```python
pod = POD(fom, centering = True, eps=1e-6)
pod.compute_basis()
pod.info()
```

```python
offline = OfflinePhase(fom, pod, x_scaler="minmax")
offline.train_rom()
offline.info()
```

```python
online = OnlinePhase(fom, pod, offline)
online.reconstruct_sol()
online.info()
```
